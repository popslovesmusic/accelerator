```

from dataclasses import dataclass
from enum import Enum
from itertools import product
import json

class Status(Enum):
    CORE="core"
    LOCAL="local"
    DERIVED="derived"
    PROJECTED="projected"

@dataclass
class Domain:
    name:str
    dof:int

DOMAINS={
 "core":Domain("core",0),
 "ordinal":Domain("ordinal",1),
 "gradient":Domain("gradient",2),
 "temporal":Domain("temporal",3),
}
print(DOMAINS)
```

```

CURRENT_DOMAIN=None

def declare_domain(name):
    global CURRENT_DOMAIN
    if name not in DOMAINS:
        raise ValueError("Unknown domain")
    CURRENT_DOMAIN=name

def require_domain():
    if CURRENT_DOMAIN is None:
        raise RuntimeError("Declare domain first.")

declare_domain("core")
print("Domain:",CURRENT_DOMAIN)
```

```

@dataclass
class Projection:
    source:str
    target:str

    def apply(self,obj):
        return {
            "image":f"{obj}@{self.target}",
            "preserved":["primitive_signature"],
            "residue":{"projection":"illustrative"}
        }

p=Projection("core","ordinal")
print(p.apply("(*|*)"))
```

```

def orientations(dof):
    return list(product((-1,1), repeat=dof))

for d in range(6):
    print(d,len(orientations(d)))
```

```

from collections import defaultdict

groups=defaultdict(list)

for o in orientations(4):
    groups[sum(o)].append(o)

for k,v in groups.items():
    if len(v)>1:
        print("Resolution",k,"examples",v[:3])
```
