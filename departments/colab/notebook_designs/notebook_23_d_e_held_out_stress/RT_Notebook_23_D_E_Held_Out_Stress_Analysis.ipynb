This notebook is the post-run documentation/analysis surface for `MPF_SIM_D_E_HELD_OUT_STRESS_001`. It does not manufacture a run. Interpretation ceiling: C2 limitation or negative evidence. No finite result may be interpreted as universal preservation, threshold derivation, injectivity, reversibility, theorem closure, or external physical validity.

```python
import json
from pathlib import Path
from collections import Counter

# In Colab, set REPO_ROOT to the mounted/cloned repository location.
REPO_ROOT = Path('/content/acellorator')
RUN_ID = 'MPF_SIM_D_E_HELD_OUT_STRESS_001'
DESIGN = REPO_ROOT / 'departments/colab/notebook_designs/notebook_23_d_e_held_out_stress'
RESULTS = REPO_ROOT / 'departments/colab/results' / RUN_ID
SPEC = DESIGN / 'experiment_spec.json'
REQUIRED = {
    'held_out_rows': RESULTS / 'held_out_rows.jsonl',
    'summary': RESULTS / 'summary.json',
    'falsification_report': RESULTS / 'falsification_report.json',
    'manifest': RESULTS / 'manifest.json',
}

def load_json(path):
    with path.open(encoding='utf-8') as handle:
        return json.load(handle)

missing = [name for name, path in REQUIRED.items() if not path.exists()]
if missing:
    analysis = {
        'run_id': RUN_ID,
        'status': 'BLOCKED_NO_EXECUTION_ARCHIVE',
        'missing_outputs': missing,
        'interpretation': 'No empirical result is available; execute the frozen Colab design and preserve its archive before analysis.',
        'claim_ceiling': 'C1_UNEXECUTED_DESIGN'
    }
else:
    with REQUIRED['held_out_rows'].open(encoding='utf-8') as handle:
        rows = [json.loads(line) for line in handle if line.strip()]
    summary = load_json(REQUIRED['summary'])
    falsification = load_json(REQUIRED['falsification_report'])
    manifest = load_json(REQUIRED['manifest'])
    analysis = {
        'run_id': RUN_ID,
        'status': 'ARCHIVE_PRESENT_REQUIRES_GOVERNED_REVIEW',
        'row_count': len(rows),
        'classification_counts': dict(Counter(row.get('classification') for row in rows)),
        'noncollapse_counts': dict(Counter(row.get('noncollapse_classification') for row in rows)),
        'replay_agreement': summary.get('replay_agreement'),
        'falsification_report': falsification,
        'manifest_keys': sorted(manifest.keys()),
        'interpretation': 'Bounded report only; preserve mismatches and counterexamples for governed review.',
        'claim_ceiling': 'C2_LIMITATION_OR_NEGATIVE_RESULT'
    }

print(json.dumps(analysis, indent=2, sort_keys=True))
```
